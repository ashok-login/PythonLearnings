### Exercise 3 - Error handling functionality

```python
class BankAccount:
    def __init__(self, account_number, owner, balance=0):
        if balance < 0:
            raise ValueError("Opening balance cannot be negative.")

        self.account_number = account_number
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <=0:
            raise ValueError("Deposit amount must be breater than zero.")
        self.balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawl amount must be greater than zero.")

        if amount > self.balance:
            raise ValueError("Insufficient funds.")
        self.balance -= amount

    def get_balance(self):
        return self.balance

# Create an account and deposit money.
account = BankAccount("ACC101", "Ashok", 1000)
account.deposit(200)
print(f"Balance after deposit: ", account.get_balance())

# Attempt a withdrawl exceeding the available balance.
try:
    account.withdraw(2000)
except ValueError as error:
    print("Withdrawl rejected: ", error)

print("Balance after rejected withdrawl: ", account.get_balance())
```

### Exercise 4 - Build an Order from Order Items

```python
class OrderItem:
    def __init__(self, product_name, unit_price, quantity):
        self.product_name = product_name
        self.unit_price = unit_price
        self.quantity = quantity

    def get_total(self):
        return self.unit_price * self.quantity

class Order:
    def __init__(self, order_id):
        self.order_id = order_id
        self.items = [] # Each order gets its own list

    def add_item(self, item):
        self.items.append(item)

    def get_total(self):
        return sum(item.get_total() for item in self.items)

# Create an order and add purchased items
order1 = Order(101)
order1.add_item(OrderItem("Keyboard", 1500, 2))
order1.add_item(OrderItem("Mouse", 700, 1))

# Create a separate, empty order
order2 = Order(102)

print("Order 101 total: ", order1.get_total())
print("Order 101 item count: ", len(order1.items))

print("Order 102 total: ", order2.get_total())
print("Order 102 item count: ", len(order2.items))
```

### Exercise 6 - xtend an Employee with Manager Details

```python
class Employee:
    def __init__(self, name, monthly_salary):
        self.name = name
        self.monthly_salary = monthly_salary

    def get_annual_salary(self):
        return self.monthly_salary * 12

class Manager(Employee):
    def __init__(self, name, monthly_salary, team_size):
        super().__init__(name, monthly_salary)
        self.team_size = team_size

    def get_team_summary(self):
        summary = f"{self.name}  manages {self.team_size} employees"
        return summary

manager = Manager("Meera", 400000, 6)
print(manager.name)
print(manager.get_annual_salary())
print(manager.get_team_summary())
```

### Hierarchical inheritance - Calculate monthly pay for developers and sales representatives, who receive different additions to their base salary.

```python
class Employee:
    def __init__(self, name, department, base_salary):
        self.name = name
        self.department = department
        self.base_salary = base_salary

    def describe(self):
        return f"{self.name} ({self.department})"

    def monthly_pay(self):
        return self.base_salary

class Developer(Employee):
    def __init__(self, name, base_salary, bonus):
        super().__init__(name, "Engineering", base_salary)
        self.bonus = bonus

    def monthly_pay(self):
        return super().monthly_pay() + self.bonus

class SalesRepresentative(Employee):
    def __init__(self, name, base_salary, commission):
        super().__init__(name, "Sales", base_salary)
        self.commission = commission

    def monthly_pay(self):
        return super().monthly_pay() + self.commission

developer = Developer("Asha", 60000, 5000)
salesperson = SalesRepresentative("Ravi", 40000, 8000)

print(f"{developer.describe()} | Pay: {developer.monthly_pay():.2f}")
print(f"{salesperson.describe()} | Pay: {salesperson.monthly_pay():.2f}")
```

### Hierarchical inheritance example

```python
from decimal import Decimal

class BankAccount:
    def __init__(self, opening_balance):
        self._balance = Decimal(opening_balance)
        if not self._balance.is_finite() or self._balance < 0:
            raise ValueError("Opening balance must be finite and non-negative.")

    def balance(self):
        return self._balance

    def _positive_amount(self, amount):
        amount = Decimal(amount)
        if not amount.is_finite() or amount <= 0:
            raise ValueError("Amount must be positive and finite.")
        return amount

    def deposit(self, amount):
        self._balance += self._positive_amount(amount)

    def available_to_withdraw(self):
        return self._balance

    def withdraw(self, amount):
        amount = self._positive_amount(amount)
        if amount > self.available_to_withdraw():
            raise ValueError("Withdrawl exceeds the avaialble amount.")
        self._balance -= amount

class SavingsAccount(BankAccount):
    def available_to_withdraw(self):
        return max(Decimal("0"), self._balance - Decimal("100"))

class CurrentAccount(BankAccount):
    def available_to_withdraw(self):
        return self._balance + Decimal("500")

savings = SavingsAccount("1200")
current = CurrentAccount("100")

savings.deposit("300")
savings.withdraw("500")
current.withdraw("400")

print(f"Savings balance: {savings.balance():.2f}")
print(f"Current balance: {current.balance():.2f}")

try:
    savings.withdraw("1000")
except ValueError as error:
    print(f"Rejected: {error}")

print(f"Savings after rejection: {savings.balance():.2f}")
```

### Abstract class example

```python
from abc import ABC, abstractmethod

class Notification(ABC):
    def format_message(self, message):
        return f"[Learning App] {message}"

    @abstractmethod
    def send(self, message):
        pass

class EmailNotification(Notification):
    def send(self, message):
        print(f"Email: {self.format_message(message)}")

class SmsNotification(Notification):
    def send(self, message):
        print(f"SMS: {self.format_message(message)}")

email = EmailNotification()
sms = SmsNotification()

email.send("Your lesson is ready.")
sms.send("Your lesson is ready.")
```

```python
from abc import ABC, abstractmethod

class PaymentProcessor(ABC):
    def __init__(self, merchant):
        self.merchant = merchant

    def pay(self, amount):
        if(amount <= 0):
            raise ValueError("Amount must be positive.")
        return f"{self.merchant} {self._charge(amount)}"

    @abstractmethod
    def _charge(self, amount):
        pass

class CardProcessor(PaymentProcessor):
    def __init__(self, merchant, last_four_digits):
        super().__init__(merchant)
        self.last_four_digits = last_four_digits

    def _charge(self, amount):
        return f"charged ₹{amount:.2f} via card ending with {self.last_four_digits}"

class WalletProcessor(PaymentProcessor):
    def __init__(self, merchant, wallet_id):
        super().__init__(merchant)
        self._wallet_id = wallet_id

    def _charge(self, amount):
        return f"charged ₹{amount:.2f} via wallet {self._wallet_id}"

card = CardProcessor("Book shop", 6000)
wallet = WalletProcessor("Book shop", "W101")

print(card.pay(500))
print(wallet.pay(300))
```

### OpenAI with Gemini base url

```python
from openai import OpenAI

from uitls import _load_env_file
from pathlib import Path
import os
from dotenv import load_dotenv

env_file = _load_env_file(Path.cwd())
load_dotenv(dotenv_path=env_file)

api_key = os.getenv("GEMINI_API_KEY")

client = OpenAI(
    api_key=api_key,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/" #Right path?
)

response = client.chat.completions.create(
    model = "gemini-2.5-flash",
    messages = [
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "How does a computer work?"}
    ],
    temperature = 0.7,
    top_p = 0.9,
    max_tokens = 500
)
print(response.choices[0].message.content)
```
