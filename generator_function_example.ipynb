### Extract error information from large log files

```python
def get_error_logs(file_path):
    with open(file_path, encoding="utf-8") as file:
        for line in file:
            if "error" in line.lower():
                yield line.rstrip()

errors = get_error_logs("E:/Delete/ErrorLog.log")

for error in errors:
    print(error)
```

### Comprehension examples

```python
squares = []
numbers = [10, 10, 15, 15, 20, 20, 25, 25, 30, 30, 35, 35]

for number in numbers:
    squares.append(number**2);
print(f"Without using comprehension: {squares}")

squares = [number**2 for number in numbers]
print(f'squares array with comprehension: {squares}')

even_numbers = [number for number in numbers if number%2 == 0]
print(f'Even numbers found using list comprehension: {even_numbers}')

unique_numbers = {number for number in numbers}
print(f'Unique numbers found using set comprehension: {unique_numbers}')

dictionary_comprehension_squares = {number: number**2 for number in numbers }
print(f'Squares as values using dictionary comprehension: {dictionary_comprehension_squares}')
```

### Higher order function: Apply different discounts to an order

```python
def regular_discount(amount):
    return amount * 0.90 # 10% discount

def premium_discount(amount):
    return amount * 0.80 # 20% discount

def apply_discount(amount, discount_function):
    return discount_function(amount)

bill_amount = 1000
regular_discount = apply_discount(bill_amount, regular_discount)
premium_discount = apply_discount(bill_amount, premium_discount)

print(f'for {bill_amount}, amount to pay after regular discount is: {regular_discount}')
print(f'for {bill_amount}, amount to pay after premium discount is: {premium_discount}')
```

### Find high-value orders for a sales review.

```python
orders = [
    {"id": 101, "amount": 1500},
    {"id": 102, "amount": 2500},
    {"id": 103, "amount": 3000}
]

def is_higher_order(order, amount):
    return order["amount"] > amount

def find_higher_orders(orders, amount, filter_condition):
    return [order
            for order in orders
            if filter_condition(order, amount)]

amount = 2000
higher_orders = find_higher_orders(orders, amount, filter_condition=is_higher_order)
print(higher_orders)
```

### Find higher-orders using lambda

```python
orders = [
    {"id": 101, "amount": 1500},
    {"id": 102, "amount": 2500},
    {"id": 103, "amount": 3000}
]

def find_higher_orders(orders, amount, filter_condition):
    return [order
            for order in orders
            if filter_condition(order, amount)]

amount = 2000
higher_orders = find_higher_orders(orders,
                    amount,
                    filter_condition=lambda order , amount: order['amount'] > amount)
print(higher_orders)
```

### One more example for lambda - Apply different discounts based on the order amount

```python
calculate_final_amount = lambda amount: (
    amount * 0.80 if amount > 5000 else
    amount * 0.90 if amount > 2000 else
    amount
)

amount = 6000
print(f'For Rs.{amount} bill, final amount to pay after '
      f'discount is: {calculate_final_amount(amount)}')
amount = 4000
print(f'For Rs.{amount} bill, final amount to pay after '
      f'discount is: {calculate_final_amount(amount)}')
amount = 2000
print(f'For Rs.{amount} bill, final amount to pay after '
      f'discount is: {calculate_final_amount(amount)}')
```

### Print prime numbers in the reange 1..50

```python
def define_prime(number):
    divisors_count = 0
    for i in range(2,number//2 + 1):
        if number % i == 0:
            divisors_count += 1
    return divisors_count == 0

prime_numbers = []
for i in range(1,51):
    if define_prime(i):
        prime_numbers.append(i)
print(prime_numbers)
```

### Program to return number of vowels, and consonants

```python
def count_vowels_consonants(text):
    text = text.lower()

    vowels_set = {
        character for character in text
        if character in "aeiou"
    }

    consonants_set = {
        character for character in text
        if "a" <= character <= "z" and character not in "aeiou"
    }

    return vowels_set, consonants_set

text = "Hello world! 123"
vowels_set, consonants_set = count_vowels_consonants(text)
print(sorted(vowels_set))
print(sorted(consonants_set))
```

### Decorator function - Log when a calculation starts and finishes

```python
def log_execution(func):
    def wrapper(*args, **kwargs):
        print('Function started')

        result = func(*args, **kwargs)

        print('Function finished')
        return result
    return wrapper
```

```python
@log_execution
def add(a, b):
    return a + b
result = add(5,10)
print(result)
```

### Python decorator to cache previously calculated results

```python
from functools import lru_cache

@lru_cache(maxsize=128)
def calculate_total(price, quantity):
    print("Calculating total...")
    return price * quantity

print(calculate_total(1000, 2))  # Calculates and saves the result.
print(calculate_total(1000, 2))  # Reuses the saved result.
print(calculate_total(1000, 3))  # Different arguments: calculates again.
```

### Python decorator for logging - Log an order-total calculation

```python
import logging
from functools import wraps

logging.basicConfig(
    level=logging.INFO,
    format="%(levelname)s: %(message)s",
    force=True, # Refresh Logging configuration in your practice notebook.
    filename="E:/Delete/application.log"
)

def log_execution(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        logging.info("Starting %s", func.__name__)

        try:
            result = func(*args, *kwargs)
        except Exception:
            logging.exception("Failed: %s", func.__name__)

            # Add a blank line, 100 equals signs, and another blank line
            logging.error("\n%s\n", "="*100)
            raise

        logging.info("Completed %s", func.__name__)
        return result

    return wrapper

@log_execution
def calculate_total(price, quantity):
    return price * quantity

try:
    # total = calculate_total(1500, 2)
    total = calculate_total(1500, None)
except TypeError:
    print("The caller handled the error.")
```
