```python
list1 = [1,4,7,8,9]

list2 = [i**2 for i in range(len(list1))] # list comprehension
list3 = [i**2 for i in list1] # list comprehension
print (list2)
print (list3)
```

### map function

```python
def square(x):
    return x**2

resultobj = map(square, [1,4,7,8,9]) # map function
list(resultobj) # returns the first value
```

### filter function

```python
resultobj = map(square, [1,4,7,8,9]) # map function
list(resultobj) # returns the first value
```

```python
def is_even(x):
    return x%2 == 0

filter_result = filter(is_even, [1,4,7,8,9]) # filter function
list(filter_result) # returns the first value
```
