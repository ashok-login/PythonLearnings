### yield - Basic Example

```python
def get_order_ids():
    yield 101
    yield 102
    yield 103

for order_id in get_order_ids():
    print(order_id)
```

```python
def read_logs(file_path):
    with open(file_path, encoding="utf-16") as file:
        for line in file:
            yield line.strip()

for log in read_logs("E:/Ashok/Notes/Notes/Links.txt"):
    print(log)
```
