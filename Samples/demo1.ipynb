```python
def process_app(*id):
    for i in id:
        if i == 2:
            print("id 2 is processed")
```

```python
process_app(1, 2, 3, 4, 5)
```

```python
def process_files(**files):
    print(type(files))
    for key, value in files.items():
        print(f"{key} : {value}")
```

```python
process_files(file1="file1.txt", file2="file2.txt", file3="file3.txt")
```

```python
def sum_numbers(*numbers):
    total = 0
    for num in numbers:
        total += num
    return total
```

```python
total = sum_numbers(10, 20, 30, 40, 50)
print(f"The sum of numbers is: {total}")
```
